```python

# Store booking details
bookings = []

# Store booked seats separately for each movie show
booked_seats = {}

# Find movie details
def get_movie(movie_name):
    for movie in movies:
        if movie["Name"] == movie_name:
            return movie
    return None


# Calculate total ticket price using NumPy
def calculate_total(price, tickets):
    total = np.array([price]) * np.array([tickets])
    return int(total[0])


# Create a new booking
def create_booking(customer_name, movie_name, tickets, seats,
                   show_date, show_time):

    movie = get_movie(movie_name)

    if movie is None:
        return None

    # Each movie, date, and time has separate seats
    show_key = (movie_name, show_date, show_time)

    if show_key not in booked_seats:
        booked_seats[show_key] = []

    # Check seat availability for this particular show
    for seat in seats:
        if seat in booked_seats[show_key]:
            return None

    total = calculate_total(movie["Price"], tickets)

    # Generate a unique Booking ID
    booking_id = random.randint(1000, 9999)

    while any(b["Booking ID"] == booking_id for b in bookings):
        booking_id = random.randint(1000, 9999)

    booking = {
        "Booking ID": booking_id,
        "Customer": customer_name,
        "Movie": movie_name,
        "Tickets": tickets,
        "Seats": seats.copy(),
        "Price": movie["Price"],
        "Total": total,
        "Show Date": show_date,
        "Show Time": show_time
    }

    bookings.append(booking)

    # Mark seats as booked for this show only
    for seat in seats:
        booked_seats[show_key].append(seat)

    return booking
```

```python


# CELL 6: Professional Movie Booking and Seat Selection Dashboard

from datetime import datetime, timedelta

def book_ticket():

    booking_window = tk.Toplevel(root)
    booking_window.title("Movie Ticket Booking")
    booking_window.geometry("780x850")
    booking_window.configure(bg="#f3f5f9")
    booking_window.resizable(False, False)

    selected_seats = []
    selected_date_var = tk.StringVar(value="")
    selected_time_var = tk.StringVar(value="")

    # ---------------- HEADER ----------------
    header = tk.Frame(booking_window, bg="#172033", height=75)
    header.pack(fill="x")
    header.pack_propagate(False)

    tk.Label(
        header,
        text="MOVIE TICKET BOOKING",
        font=("Arial", 20, "bold"),
        bg="#172033",
        fg="white"
    ).pack(pady=20)

    # ---------------- MAIN CONTENT ----------------
    content = tk.Frame(booking_window, bg="#f3f5f9")
    content.pack(fill="both", expand=True, padx=22, pady=12)

    # Customer name
    tk.Label(
        content,
        text="Customer Name",
        font=("Arial", 10, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(anchor="w")

    customer_entry = tk.Entry(
        content,
        font=("Arial", 11),
        relief="solid",
        bd=1
    )
    customer_entry.pack(fill="x", pady=(4, 12), ipady=5)

    # ---------------- MOVIE SELECTION ----------------
    tk.Label(
        content,
        text="1. Select Movie",
        font=("Arial", 12, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(anchor="w", pady=(0, 5))

    movie_names = [movie["Name"] for movie in movies]
    movie_var = tk.StringVar(value=movie_names[0])

    movie_menu = tk.OptionMenu(
        content, movie_var, *movie_names
    )
    movie_menu.config(
        font=("Arial", 10),
        bg="white",
        width=25,
        relief="solid"
    )
    movie_menu.pack(anchor="w", pady=(0, 12))

    # ---------------- DATE SELECTION ----------------
    tk.Label(
        content,
        text="2. Choose Your Show Date",
        font=("Arial", 12, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(anchor="w", pady=(0, 7))

    # Four available dates: today and the following three days
    available_dates = [
        datetime.now().date() + timedelta(days=i)
        for i in range(4)
    ]

    date_frame = tk.Frame(content, bg="#f3f5f9")
    date_frame.pack(fill="x", pady=(0, 12))

    date_buttons = {}

    def refresh_seats():
        movie_name = movie_var.get()
        show_date = selected_date_var.get()
        show_time = selected_time_var.get()

        show_key = (movie_name, show_date, show_time)
        occupied_seats = booked_seats.get(show_key, [])

        for seat, button in seat_buttons.items():

            if seat in occupied_seats:
                button.config(
                    bg="#dc2626",
                    fg="white",
                    state="disabled"
                )

            elif not show_date or not show_time:
                button.config(
                    bg="#e5e7eb",
                    fg="#6b7280",
                    state="disabled"
                )

            elif seat in selected_seats:
                button.config(
                    bg="#16a34a",
                    fg="white",
                    state="normal"
                )

            else:
                button.config(
                    bg="white",
                    fg="#172033",
                    state="normal"
                )

    def choose_date(date_text):
        selected_date_var.set(date_text)
        selected_seats.clear()

        for date_text_button, button in date_buttons.items():
            if date_text_button == date_text:
                button.config(
                    bg="#2563eb",
                    fg="white",
                    relief="sunken"
                )
            else:
                button.config(
                    bg="white",
                    fg="#172033",
                    relief="raised"
                )

        refresh_seats()

    for index, date_value in enumerate(available_dates):

        date_text = date_value.strftime("%d-%m-%Y")
        day_text = date_value.strftime("%a")
        button_text = (
            f"{day_text}\n"
            f"{date_value.strftime('%d %b')}"
        )

        date_button = tk.Button(
            date_frame,
            text=button_text,
            font=("Arial", 10, "bold"),
            width=11,
            height=2,
            bg="white",
            fg="#172033",
            relief="raised",
            bd=1,
            command=lambda d=date_text: choose_date(d)
        )

        date_button.grid(
            row=0,
            column=index,
            padx=4,
            sticky="ew"
        )

        date_buttons[date_text] = date_button
        date_frame.grid_columnconfigure(index, weight=1)

    # ---------------- SHOW TIME ----------------
    tk.Label(
        content,
        text="3. Select Show Time",
        font=("Arial", 12, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(anchor="w", pady=(0, 5))

    time_frame = tk.Frame(content, bg="#f3f5f9")
    time_frame.pack(fill="x", pady=(0, 12))

    show_times = [
        "10:00 AM",
        "01:00 PM",
        "04:00 PM",
        "06:00 PM",
        "09:00 PM"
    ]

    time_buttons = {}

    def choose_time(time_text):
        selected_time_var.set(time_text)
        selected_seats.clear()

        for button_time, button in time_buttons.items():
            if button_time == time_text:
                button.config(bg="#7c3aed", fg="white")
            else:
                button.config(bg="white", fg="#172033")

        refresh_seats()

    for index, show_time in enumerate(show_times):

        time_button = tk.Button(
            time_frame,
            text=show_time,
            font=("Arial", 9, "bold"),
            bg="white",
            fg="#172033",
            width=11,
            command=lambda t=show_time: choose_time(t)
        )

        time_button.grid(
            row=0,
            column=index,
            padx=3,
            pady=2,
            sticky="ew"
        )

        time_buttons[show_time] = time_button
        time_frame.grid_columnconfigure(index, weight=1)

    # ---------------- TICKET COUNT ----------------
    ticket_frame = tk.Frame(content, bg="#f3f5f9")
    ticket_frame.pack(fill="x", pady=(0, 12))

    tk.Label(
        ticket_frame,
        text="Number of Tickets:",
        font=("Arial", 11, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(side="left")

    tickets_entry = tk.Entry(
        ticket_frame,
        font=("Arial", 11),
        width=8,
        relief="solid",
        bd=1
    )
    tickets_entry.pack(side="left", padx=10, ipady=3)

    # ---------------- SEAT DASHBOARD ----------------
    tk.Label(
        content,
        text="4. Select Your Seats",
        font=("Arial", 12, "bold"),
        bg="#f3f5f9",
        fg="#172033"
    ).pack(anchor="w", pady=(0, 8))

    seat_panel = tk.Frame(
        content,
        bg="white",
        bd=1,
        relief="solid"
    )
    seat_panel.pack(fill="x", pady=(0, 8))

    tk.Label(
        seat_panel,
        text="SCREEN",
        font=("Arial", 10, "bold"),
        bg="#dbeafe",
        fg="#1d4ed8",
        pady=7
    ).pack(fill="x", padx=35, pady=(12, 15))

    seat_buttons = {}

    seat_grid = tk.Frame(seat_panel, bg="white")
    seat_grid.pack(pady=(0, 10))

    # Four rows, five seats per row
    for row_index in range(4):

        tk.Label(
            seat_grid,
            text=chr(65 + row_index),
            font=("Arial", 10, "bold"),
            bg="white",
            fg="#64748b",
            width=2
        ).grid(row=row_index, column=0, padx=4)

        for col_index in range(5):

            seat_number = row_index * 5 + col_index + 1
            seat = f"S{seat_number}"

            seat_button = tk.Button(
                seat_grid,
                text=seat,
                font=("Arial", 9, "bold"),
                width=7,
                height=1,
                bg="#e5e7eb",
                fg="#6b7280",
                state="disabled",
                command=lambda s=seat: toggle_seat(s)
            )

            seat_button.grid(
                row=row_index,
                column=col_index + 1,
                padx=4,
                pady=5
            )

            seat_buttons[seat] = seat_button

    # Seat color legend
    legend = tk.Frame(seat_panel, bg="white")
    legend.pack(pady=(0, 12))

    for color, label_text in [
        ("#e5e7eb", "Available"),
        ("#16a34a", "Selected"),
        ("#dc2626", "Booked")
    ]:
        tk.Label(
            legend,
            text=f"  {label_text}  ",
            bg=color,
            fg="white" if color != "#e5e7eb" else "#172033",
            font=("Arial", 9, "bold"),
            padx=4,
            pady=3
        ).pack(side="left", padx=5)

    def toggle_seat(seat):
        show_date = selected_date_var.get()
        show_time = selected_time_var.get()

        if not show_date or not show_time:
            messagebox.showwarning(
                "Select Show",
                "Please select a date and show time first."
            )
            return

        show_key = (movie_var.get(), show_date, show_time)

        if seat in booked_seats.get(show_key, []):
            messagebox.showwarning(
                "Seat Booked",
                f"{seat} is already booked for this show."
            )
            refresh_seats()
            return

        if seat in selected_seats:
            selected_seats.remove(seat)
        else:
            # Do not select more seats than the ticket count
            try:
                ticket_count = int(tickets_entry.get())
                if ticket_count <= 0:
                    raise ValueError
            except ValueError:
                messagebox.showwarning(
                    "Ticket Count",
                    "Enter the number of tickets before selecting seats."
                )
                return

            if len(selected_seats) >= ticket_count:
                messagebox.showwarning(
                    "Seat Limit",
                    f"You can select only {ticket_count} seat(s)."
                )
                return

            selected_seats.append(seat)

        refresh_seats()

    # Reset selected seats when movie changes
    def movie_changed(*args):
        selected_seats.clear()
        refresh_seats()

    movie_var.trace_add("write", movie_changed)

    # ---------------- CONFIRM BOOKING ----------------
    def confirm_booking():

        customer_name = customer_entry.get().strip()
        movie_name = movie_var.get()
        show_date = selected_date_var.get()
        show_time = selected_time_var.get()

        if not customer_name:
            messagebox.showerror(
                "Missing Name",
                "Please enter the customer name."
            )
            return

        if not show_date:
            messagebox.showerror(
                "Missing Date",
                "Please select one of the four available dates."
            )
            return

        if not show_time:
            messagebox.showerror(
                "Missing Time",
                "Please select a show time."
            )
            return

        try:
            tickets = int(tickets_entry.get())
            if tickets <= 0:
                raise ValueError
        except ValueError:
            messagebox.showerror(
                "Invalid Tickets",
                "Please enter a valid number of tickets."
            )
            return

        if len(selected_seats) != tickets:
            messagebox.showerror(
                "Seat Selection",
                f"Select exactly {tickets} seat(s)."
            )
            return

        # Check the selected date is not in the past
        selected_date = datetime.strptime(
            show_date, "%d-%m-%Y"
        ).date()

        if selected_date < datetime.now().date():
            messagebox.showerror(
                "Invalid Date",
                "You cannot book a show in the past."
            )
            return

        seats_to_book = selected_seats.copy()

        booking = create_booking(
            customer_name,
            movie_name,
            tickets,
            seats_to_book,
            show_date,
            show_time
        )

        if booking is None:
            messagebox.showerror(
                "Booking Error",
                "One or more selected seats are already booked. "
                "Please choose other seats."
            )
            refresh_seats()
            return

        movie = get_movie(movie_name)

        receipt = (
            "BOOKING CONFIRMED!\n\n"
            f"Booking ID: {booking['Booking ID']}\n"
            f"Customer: {customer_name}\n"
            f"Movie: {movie_name}\n"
            f"Show Date: {show_date}\n"
            f"Show Time: {show_time}\n"
            f"Seats: {', '.join(seats_to_book)}\n"
            f"Tickets: {tickets}\n"
            f"Price per Ticket: ₹{movie['Price']}\n"
            f"Total Amount: ₹{booking['Total']}"
        )

        messagebox.showinfo("Booking Receipt", receipt)
        booking_window.destroy()

    tk.Button(
        content,
        text="CONFIRM BOOKING",
        font=("Arial", 12, "bold"),
        bg="#2563eb",
        fg="white",
        activebackground="#1d4ed8",
        activeforeground="white",
        relief="flat",
        cursor="hand2",
        pady=10,
        command=confirm_booking
    ).pack(fill="x", pady=(5, 10))

    refresh_seats()
```

```python

def cancel_booking():

    # Check whether bookings exist
    if len(bookings) == 0:
        messagebox.showinfo(
            "Cancel Booking",
            "No bookings available to cancel."
        )
        return

    # Create cancellation window
    cancel_window = tk.Toplevel(root)
    cancel_window.title("Cancel Booking")
    cancel_window.geometry("500x350")
    cancel_window.configure(bg="white")

    # Heading
    heading = tk.Label(
        cancel_window,
        text="CANCEL BOOKING",
        font=("Arial", 22, "bold"),
        bg="white",
        fg="#172033"
    )
    heading.pack(pady=25)

    # Instructions
    tk.Label(
        cancel_window,
        text="Enter your Booking ID",
        font=("Arial", 12, "bold"),
        bg="white",
        fg="#374151"
    ).pack(pady=5)

    # Booking ID input
    booking_entry = tk.Entry(
        cancel_window,
        width=30,
        font=("Arial", 12)
    )
    booking_entry.pack(pady=10)

    # Cancel function
    def confirm_cancel():

        booking_text = booking_entry.get().strip()

        # Validate Booking ID
        if booking_text == "":
            messagebox.showwarning(
                "Missing Information",
                "Please enter your Booking ID.",
                parent=cancel_window
            )
            return

        try:
            booking_id = int(booking_text)
        except ValueError:
            messagebox.showerror(
                "Invalid Booking ID",
                "Please enter a valid numeric Booking ID.",
                parent=cancel_window
            )
            return

        # Find the booking
        booking_found = None

        for booking in bookings:
            if booking["Booking ID"] == booking_id:
                booking_found = booking
                break

        # If booking does not exist
        if booking_found is None:
            messagebox.showerror(
                "Booking Not Found",
                "No booking found with this Booking ID.",
                parent=cancel_window
            )
            return

        # Ask for confirmation
        answer = messagebox.askyesno(
            "Confirm Cancellation",
            "Are you sure you want to cancel this booking?\n\n"
            f"Movie: {booking_found['Movie']}\n"
            f"Seats: {', '.join(booking_found['Seats'])}\n"
            f"Total: Rs. {booking_found['Total']}",
            parent=cancel_window
        )

        if not answer:
            return

        # Get the movie associated with this booking
        movie_name = booking_found["Movie"]

        # Release seats for this movie only
        if movie_name in booked_seats:

            for seat in booking_found["Seats"]:

                if seat in booked_seats[movie_name]:
                    booked_seats[movie_name].remove(seat)

        # Remove the booking record
        bookings.remove(booking_found)

        # Show success message
        messagebox.showinfo(
            "Cancellation Successful",
            f"Booking ID: {booking_id}\n"
            f"Movie: {movie_name}\n"
            f"Released seats: "
            f"{', '.join(booking_found['Seats'])}\n\n"
            "Your booking has been cancelled.",
            parent=cancel_window
        )

        # Close cancellation window
        cancel_window.destroy()

    # Cancel Booking button
    cancel_button = tk.Button(
        cancel_window,
        text="CANCEL BOOKING",
        font=("Arial", 12, "bold"),
        bg="#dc2626",
        fg="white",
        activebackground="#b91c1c",
        activeforeground="white",
        bd=0,
        width=22,
        height=2,
        command=confirm_cancel
    )
    cancel_button.pack(pady=20)
```
